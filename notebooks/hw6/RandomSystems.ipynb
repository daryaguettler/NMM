## Cumulants

To get the cumulants, we need to:
Write $\varphi(k) = 1 + u$ 

We define u as:
$$u = \frac{ik}{1!}\langle x \rangle + \frac{(ik)^2}{2!}\langle x^2 \rangle + \frac{(ik)^3}{3!}\langle x^3 \rangle + \cdots$$

Then use the taylor expansion $\log(1+u) = u - \tfrac{u^2}{2} + \tfrac{u^3}{3} - \cdots$ and collect terms order by order in $k$.

We srtart with teh first one:

The only contribution at order $k$ comes from the first term of $u$:

$$\log\varphi(k)\big|_{\mathcal{O}(k)} = ik\langle x\rangle$$

$$C_1 = \langle x \rangle$$

The first cumulant is the average! 

For $k_2$
- The $k^2$ term in $u$: $\quad \dfrac{(ik)^2}{2}\langle x^2\rangle = -\dfrac{k^2}{2}\langle x^2\rangle$
- The $k^2$ term in $-\tfrac{u^2}{2}$: $\quad -\dfrac{1}{2}(ik)^2\langle x\rangle^2 = +\dfrac{k^2}{2}\langle x\rangle^2$

we can sum those terms to get the second constribution:

$$\log\varphi(k)\big|_{\mathcal{O}(k^2)} = -\frac{k^2}{2}\bigl(\langle x^2\rangle - \langle x\rangle^2\bigr)$$


$$C_2 = \langle x^2 \rangle - \langle x \rangle^2 = \operatorname{Var}(x) = \sigma^2$$

The second cumulant is the variance

1. The $k^3$ term in $u$: $\quad \dfrac{(ik)^3}{6}\langle x^3\rangle = -\dfrac{ik^3}{6}\langle x^3\rangle$

2. The $k^3$ term in $-\tfrac{u^2}{2}$:
$$-\frac{1}{2}\cdot 2\cdot(ik\langle x\rangle)\cdot\left(-\frac{k^2}{2}\langle x^2\rangle\right) = +\frac{ik^3}{2}\langle x\rangle\langle x^2\rangle$$

3. The $k^3$ term in $+\tfrac{u^3}{3}$: $\quad \dfrac{1}{3}(ik)^3\langle x\rangle^3 = -\dfrac{ik^3}{3}\langle x\rangle^3$

We can like for $k_2$ sum these 3 terms:
$$\log\varphi(k)\big|_{\mathcal{O}(k^3)} = -\frac{ik^3}{6}\Bigl(\langle x^3\rangle - 3\langle x\rangle\langle x^2\rangle + 2\langle x\rangle^3\Bigr)$$

$\dfrac{(ik)^3}{3!}C_3 = -\dfrac{ik^3}{6}C_3$:

$$C_3 = \langle x^3\rangle - 3\langle x\rangle\langle x^2\rangle + 2\langle x\rangle^3$$

This repesents the skew of the distribution

### Gaussian Distribution


Consider the Gaussian (normal) distribution:

$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\, e^{-(x-\bar{x})^2/2\sigma^2}$$
$$\varphi(k) = \langle e^{ikx}\rangle = \int_{-\infty}^{\infty} e^{ikx} \cdot \frac{1}{\sqrt{2\pi\sigma^2}}\, e^{-(x-\bar{x})^2/2\sigma^2}\, dx$$

Complete the square in the exponent. Writing $y = x - \bar{x}$:

$$ikx - \frac{(x-\bar{x})^2}{2\sigma^2} = ik\bar{x} + iky - \frac{y^2}{2\sigma^2}
= ik\bar{x} - \frac{1}{2\sigma^2}\left(y - ik\sigma^2\right)^2 - \frac{k^2\sigma^2}{2}$$

The Gaussian integral over the shifted variable gives 1, so:

$$\boxed{\varphi(k) = e^{ik\bar{x} - k^2\sigma^2/2}}$$


$$\log\varphi(k) = ik\bar{x} - \frac{k^2\sigma^2}{2}$$

THe series terminates.

$$ik C_1 + \frac{(ik)^2}{2}C_2 + \frac{(ik)^3}{6}C_3 + \cdots = ik\bar{x} - \frac{k^2\sigma^2}{2} + 0 + \cdots$$

Matching order by order:

$$C_1 = \bar{x}, \qquad C_2 = \sigma^2, \qquad C_3 = 0$$

All cumulants $C_n = 0$ for $n \geq 3$ for a Gaussian. This is the defining property of the Gaussian distribution — it is completely characterised by its mean and variance. Any non-zero higher cumulant would mean its not gaussian anymore

## Differential Elements

### Area of differential Elements

The area of a parallelogram with edges $\vec{a}$ and $\vec{b}$ equals the magnitude of their cross product. In 2D, for $\vec{a} = (a_1, a_2)$ and $\vec{b} = (b_1, b_2)$:

$$\text{Area} = |a_1 b_2 - a_2 b_1|$$

we can consider the infinitesimal rectangle with corners at $\vec{x}$, $\vec{x} + dx_1\hat{e}_1$, and $\vec{x} + dx_2\hat{e}_2$. Under $\vec{y}(\vec{x})$, the two edge vectors map to:

$$\vec{a} = \vec{y}(\vec{x} + dx_1\hat{e}_1) - \vec{y}(\vec{x}) \approx \frac{\partial \vec{y}}{\partial x_1}\, dx_1 = \begin{pmatrix}\partial y_1/\partial x_1 \\ \partial y_2/\partial x_1\end{pmatrix}dx_1$$

$$\vec{b} = \vec{y}(\vec{x} + dx_2\hat{e}_2) - \vec{y}(\vec{x}) \approx \frac{\partial \vec{y}}{\partial x_2}\, dx_2 = \begin{pmatrix}\partial y_1/\partial x_2 \\ \partial y_2/\partial x_2\end{pmatrix}dx_2$$

Therefore, $$dA_{\vec{y}} = \left|\frac{\partial y_1}{\partial x_1}\frac{\partial y_2}{\partial x_2} - \frac{\partial y_1}{\partial x_2}\frac{\partial y_2}{\partial x_1}\right| dx_1\, dx_2$$

The quantity in the absolute value is determinant of the jacobian:

$$J = \frac{\partial(y_1, y_2)}{\partial(x_1, x_2)} = \det\begin{pmatrix} \partial y_1/\partial x_1 & \partial y_1/\partial x_2 \\ \partial y_2/\partial x_1 & \partial y_2/\partial x_2 \end{pmatrix}$$

Therefore:

$$dy_1\, dy_2 = \left|\frac{\partial(y_1,y_2)}{\partial(x_1,x_2)}\right| dx_1\, dx_2$$


Probability is conserved under a change of variables and the probability in a region cannot depend on which coordinates we use to describe it:

$$p(\vec{x})\, dx_1\, dx_2 = p(\vec{y})\, dy_1\, dy_2$$

Using the Jacobian result above:

$$p(\vec{y}) = p(\vec{x})\left|\frac{\partial(y_1,y_2)}{\partial(x_1,x_2)}\right|^{-1}$$

where $\vec{x}$ on the right is understood to be expressed in terms of $\vec{y}$ via the inverse transformation.


### 2.2 

Define the transformation:

$$y_1 = \sqrt{-2\ln x_1}\,\sin(x_2), \qquad y_2 = \sqrt{-2\ln x_1}\,\cos(x_2)$$

where $x_1 \in (0,1]$ and $x_2 \in [0, 2\pi)$, with $p(x_1, x_2) = \dfrac{1}{2\pi}$

We want to find $p(y_1, y_2)$.

Note that $y_1^2 + y_2^2 = -2\ln x_1(\sin^2 x_2 + \cos^2 x_2) = -2\ln x_1$, so:

$$x_1 = e^{-(y_1^2 + y_2^2)/2}, \qquad x_2 = \arctan\!\left(\frac{y_1}{y_2}\right)$$

We cn find the jacobian of the forward transform and then invert it. therefore, we let $r^2 = y_1^2 + y_2^2 = -2\ln x_1$, so $\partial r^2/\partial x_1 = -2/x_1$. 

$$\frac{\partial y_1}{\partial x_1} = \frac{-1}{x_1 r}\sin x_2, \qquad \frac{\partial y_1}{\partial x_2} = r\cos x_2$$

$$\frac{\partial y_2}{\partial x_1} = \frac{-1}{x_1 r}\cos x_2, \qquad \frac{\partial y_2}{\partial x_2} = -r\sin x_2$$

The Jacobian determinant is:

$$J = \frac{\partial y_1}{\partial x_1}\frac{\partial y_2}{\partial x_2} - \frac{\partial y_1}{\partial x_2}\frac{\partial y_2}{\partial x_1}
= \frac{1}{x_1 r}\sin x_2\cdot r\sin x_2 + r\cos x_2 \cdot \frac{1}{x_1 r}\cos x_2
= \frac{1}{x_1}$$

So $|J| = 1/x_1 = e^{(y_1^2+y_2^2)/2}$, and therefore $|J|^{-1} = x_1 = e^{-(y_1^2+y_2^2)/2}$.

We can then transform the distribution:
$$p(y_1, y_2) = p(x_1, x_2)\,|J|^{-1} = \frac{1}{2\pi}\cdot e^{-(y_1^2+y_2^2)/2}$$

$$p(y_1, y_2) = \frac{1}{\sqrt{2\pi}}e^{-y_1^2/2}\cdot\frac{1}{\sqrt{2\pi}}e^{-y_2^2/2}$$

The joint distribution factorises into two independent standard Gaussians $\mathcal{N}(0,1)$

### 2.3 - Writing a uniform random number generator

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

rng = np.random.default_rng(seed=42)
N = 1_000_000
x1 = rng.uniform(0, 1, size=N)   
x2 = rng.uniform(0, 2*np.pi, size=N)  
r   = np.sqrt(-2 * np.log(x1)) 
y1  = r * np.sin(x2)
y2  = r * np.cos(x2)

print(f"Generated {N:,} pairs")
print(f"Sample mean of y1: {y1.mean():.6f}")
print(f"Sample std  of y1: {y1.std():.6f}")
```

```python
def cumulants(samples):
    m1 = np.mean(samples)
    m2 = np.mean(samples**2)
    m3 = np.mean(samples**3)
    C1 = m1
    C2 = m2 - m1**2
    C3 = m3 - 3*m1*m2 + 2*m1**3
    return C1, C2, C3

C1_y1, C2_y1, C3_y1 = cumulants(y1)
C1_y2, C2_y2, C3_y2 = cumulants(y2)

print("Cumulants y1:")
print(f"  C1 = {C1_y1:+.6f}")
print(f"  C2 = {C2_y1:+.6f}")
print(f"  C3 = {C3_y1:+.6f}")
print()
print("Cumulants y2:")
print(f"  C1 = {C1_y2:+.6f}")
print(f"  C2 = {C2_y2:+.6f}")
print(f"  C3 = {C3_y2:+.6f}")
```

We can see that we get very close to the normal distribution mean and variance!

## LFSR

###  Order 4 maximal LFSR bit sequence

For $n = 4$, the polynomials over $\mathrm{GF}(2)$ are:

$$f(x) = x^4 + x + 1$$
$$f(x) = x^4 + x^3 + 1$$

We use $f(x) = x^4 + x + 1$, which gives feedback:

$$s_{\text{new}} = s_4 \oplus s_1$$

where $s_1$ is the right bit and $s_4$ is the left bit.

Starting from the seed state $[1, 0, 0, 0]$ (left = $s_4$, right = $s_1$), at each step:
1. The feedback bit is $s_4 \oplus s_1$
2. Shift right: the old $s_1$ is the output bit
3. The feedback bit enters at the left as the new $s_4$

The output bit sequence has period $2^4 - 1 = \mathbf{15}$.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.gridspec import GridSpec

def lfsr(seed: list[int], taps: list[int], n_steps: int):
    n = len(seed)
    state = list(seed)
    bits, states = [], []
    for _ in range(n_steps):
        out = state[-1]
        bits.append(out)
        fb = 0
        for p in taps:
            fb ^= state[n - p]
        state = [fb] + state[:-1]
        states.append(list(state))
    return bits, states


n_order = 4
taps4   = [4, 1]
seed4   = [1, 0, 0, 0]
period4 = 2**n_order - 1

bits4, states4 = lfsr(seed4, taps4, period4)


print(f"{'Step':>4}  {'State (s4 s3 s2 s1)':>20}  {'Output bit':>11}")
for i, (s, b) in enumerate(zip(states4, bits4)):
    prev = seed4 if i == 0 else states4[i-1]
    print(f"{i+1:>4}  {str(prev):>20}  {b:>11}")
print("Output sequence: ", ''.join(str(b) for b in bits4))
```

### LFSR clock rate

A maximal $n$-bit LFSR running at clock rate $f$ produces a sequence that repeats after:

$$T_{\text{repeat}} = \frac{2^n - 1}{f} \approx \frac{2^n}{f}$$

We want $T_{\text{repeat}} \geq T_{\text{universe}} \approx 10^{10}$ years at $f = 1\,\mathrm{GHz} = 10^9\,\mathrm{Hz}$.

Therefore we can solve for n:

$$\frac{2^n}{10^9\,\mathrm{Hz}} \geq 10^{10}\,\mathrm{yr} \times 3.156 \times 10^7\,\mathrm{s\,yr^{-1}}$$

$$2^n \geq 10^9 \times 10^{10} \times 3.156 \times 10^7 = 3.156 \times 10^{26}$$

$$n \geq \log_2\!\left(3.156 \times 10^{26}\right) = \frac{\ln(3.156 \times 10^{26})}{\ln 2}$$

```python
import math
f_clock   = 1e9                         # Hz  (1 GHz)
secs_per_year = 365.25 * 24 * 3600     # s yr⁻¹
age_universe  = 1e10 * secs_per_year   # seconds

print(f"Clock f      : {f_clock:.2e} Hz")
print(f"Age of universe      : {age_universe:.4e} s  ({1e10:.0e} yr)")
print()
cycles_needed = f_clock * age_universe
print(f"Cycles needed: {cycles_needed:.4e}")
print()
n_min = math.ceil(math.log2(cycles_needed + 1))
T_repeat = (2**n_min - 1) / f_clock / secs_per_year

print(f"Minimum register length n = {n_min} bits")
print(f"  Period  = 2^{n_min} - 1 = {2**n_min - 1:.6e} clock cycles")
print(f"  Repeat time = {T_repeat:.4e} years")
print()
T_short = (2**(n_min-1) - 1) / f_clock / secs_per_year
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ns = np.arange(1, 120)
repeat_years = (2.0**ns - 1) / f_clock / secs_per_year

ax.semilogy(ns, repeat_years, color='steelblue', lw=2, label='Repeat time')
ax.axhline(1e10, color='red', lw=1.5, ls='--', label='Age of universe ($10^{10}$ yr)')
ax.axvline(n_min, color='darkorange', lw=1.5, ls=':', label=f'$n = {n_min}$ (minimum)')
ax.scatter([n_min], [T_repeat], color='darkorange', zorder=5, s=60)

ax.set_xlabel('Register length $n$ (bits)', fontsize=12)
ax.set_ylabel('Repeat time (years)', fontsize=12)
ax.set_title('LFSR repeat time vs register length at $f = 1$ GHz', fontsize=12)
ax.legend(fontsize=10)
ax.set_xlim(1, 115)
ax.grid(which='both', ls=':', alpha=0.4)

plt.tight_layout()
plt.show()
```
