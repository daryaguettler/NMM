## 1.1 — Downhill simplex in 2D

We set the stopping criteria:
 $\varepsilon = 10^{-8}$
 
$$\sigma(f(x_0), \ldots, f(x_n)) < 10^{-8}$$

Starting point: $x = y = -1$. Global minimum: $(1, 1)$.

```python
def nelder_mead_2d(func, x0, tol=1e-8, max_iter=10000):
    alpha, gamma, rho, sigma = 1.0, 2.0, 0.5, 0.5
    n = len(x0)

    simplex = [np.array(x0, dtype=float)]
    for i in range(n):
        v = np.array(x0, dtype=float)
        v[i] += 0.5
        simplex.append(v)
    feval = 0
    f = [func(v) for v in simplex]
    feval += len(simplex)
    centroid_history = []
    for iteration in range(max_iter):
        # Sort
        order = np.argsort(f)
        simplex = [simplex[i] for i in order]
        f      = [f[i]       for i in order]

        centroid_history.append(simplex[0].copy())
        if np.std(f) < tol:
            break
        centroid = np.mean(simplex[:-1], axis=0)
        x_r = centroid + alpha * (centroid - simplex[-1])
        f_r = func(x_r); feval += 1

        if f[0] <= f_r < f[-2]:
            simplex[-1], f[-1] = x_r, f_r
        elif f_r < f[0]:
            x_e = centroid + gamma * (x_r - centroid)
            f_e = func(x_e); feval += 1
            if f_e < f_r:
                simplex[-1], f[-1] = x_e, f_e
            else:
                simplex[-1], f[-1] = x_r, f_r
        else:
            x_c = centroid + rho * (simplex[-1] - centroid)
            f_c = func(x_c); feval += 1
            if f_c < f[-1]:
                simplex[-1], f[-1] = x_c, f_c
            else:
                # Shrink
                simplex = [simplex[0]] + [
                    simplex[0] + sigma * (v - simplex[0])
                    for v in simplex[1:]
                ]
                f = [func(v) for v in simplex]; feval += n

    return simplex[0], iteration + 1, feval, np.array(centroid_history)


f2d = lambda v: rosenbrock_2d(v[0], v[1])
x0_2d = [-1.0, -1.0]

best_2d, iters_2d, feval_2d, path_2d = nelder_mead_2d(f2d, x0_2d)

print(f"output:")
print(f"  Best point : x={best_2d[0]:.6f}, y={best_2d[1]:.6f}")
print(f"  f(x*)      : {f2d(best_2d):.2e}")
print(f"  Iterations : {iters_2d}")
print(f"  Func evals : {feval_2d}")
```

## 1.2 — Downhill simplex in 10D

$$f(\mathbf{x}) = \sum_{i=1}^{D-1} \left[(1 - x_{i})^2 + 100(x_{i+1} - x_i^2)^2\right]$$

Global minimum: $\mathbf{x}^* = (1, 1, \ldots, 1)$, $f^* = 0$.

## 1.3 — Gradient Descent

$$\mathbf{x}_{k+1} = \mathbf{x}_k - \alpha_k \nabla f(\mathbf{x}_k)$$

where $\alpha_k$ is chosen so that $f(\mathbf{x}_k - \alpha_k \nabla f) \leq f(\mathbf{x}_k) - c\, \alpha_k \|\nabla f\|^2$.

Stopping condition: $\|\nabla f\| < 10^{-6}$.


```python
def rosenbrock_grad_2d(v):
    x, y = v
    df_dx = -2*(1 - x) - 400*x*(y - x**2)
    df_dy =  200*(y - x**2)
    return np.array([df_dx, df_dy])

def rosenbrock_grad_nd(x):
    x = np.asarray(x, dtype=float)
    g = np.zeros_like(x)
    g[:-1] += -2*(1 - x[:-1]) - 400*x[:-1]*(x[1:] - x[:-1]**2)
    g[1:]  +=  200*(x[1:] - x[:-1]**2)
    return g

def gradient_descent(func, grad, x0, tol=1e-6, max_iter=100000,
                     alpha0=1e-3, c=1e-4, beta=0.5):
    x = np.array(x0, dtype=float)
    path = [x.copy()]
    feval = 1
    f_cur = func(x)
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        alpha = alpha0
        for _ in range(50):
            x_new = x - alpha * g
            f_new = func(x_new); feval += 1
            if f_new <= f_cur - c * alpha * np.dot(g, g):
                break
            alpha *= beta

        x = x_new
        f_cur = f_new
        path.append(x.copy())

    return x, k + 1, feval, np.array(path)

best_gd2d, iters_gd2d, feval_gd2d, path_gd2d = gradient_descent(
    f2d, rosenbrock_grad_2d, [-1.0, -1.0]
)
print("2D gd")
print(f"  Best: {best_gd2d.round(5)},  f={f2d(best_gd2d):.2e}")
print(f"  Iterations: {iters_gd2d},  Fevals: {feval_gd2d}")
rosen_nd = rosenbrock_nd
best_gd10d, iters_gd10d, feval_gd10d, path_gd10d = gradient_descent(
    rosen_nd, rosenbrock_grad_nd, np.full(10, -1.0)
)
print("10D gd")
print(f"  Best (first 4): {best_gd10d[:4].round(5)},  f={rosen_nd(best_gd10d):.2e}")
print(f"  Iterations: {iters_gd10d},  Fevals: {feval_gd10d}")
```

## 1.4 — Conjugate Gradient

**Conjugate Gradient (CG)** improves on steepest descent by choosing successive search directions that are mutually conjugate with respect to the Hessian.  

$$\beta_k^{\text{FR}} = \frac{\|\nabla f_{k+1}\|^2}{\|\nabla f_k\|^2}, \qquad \mathbf{d}_{k+1} = -\nabla f_{k+1} + \beta_k \mathbf{d}_k$$

with a reset to steepest descent whenever $\beta < 0$ (Polak–Ribière fallback).  
Each CG step uses a backtracking line search satisfying the Wolfe conditions.  


Stopping condition: $\|\nabla f\| < 10^{-6}$.

```python
def conjugate_gradient(func, grad, x0, tol=1e-6, max_iter=20000):
    x = np.array(x0, dtype=float)
    g = grad(x)
    d = -g.copy()
    path = [x.copy()]
    feval = 1
    f_cur = func(x)

    for k in range(max_iter):
        if np.linalg.norm(g) < tol:
            break

        alpha = 1.0
        slope  = np.dot(g, d)
        for _ in range(60):
            x_new  = x + alpha * d
            f_new  = func(x_new); feval += 1
            if f_new <= f_cur + 1e-4 * alpha * slope:
                break
            alpha *= 0.5

        x     = x_new
        f_cur = f_new
        g_new = grad(x)
        beta = np.dot(g_new, g_new) / (np.dot(g, g) + 1e-300)
        beta = max(beta, 0.0)   # Polak-Ribière reset if negative

        d = -g_new + beta * d
        g = g_new
        path.append(x.copy())

    return x, k + 1, feval, np.array(path)

best_cg2d, iters_cg2d, feval_cg2d, path_cg2d = conjugate_gradient(
    f2d, rosenbrock_grad_2d, [-1.0, -1.0]
)
print("2D Conjugate Gradient")
print(f"  Best: {best_cg2d.round(5)},  f={f2d(best_cg2d):.2e}")
print(f"  Iterations: {iters_cg2d},  Fevals: {feval_cg2d}")
best_cg10d, iters_cg10d, feval_cg10d, path_cg10d = conjugate_gradient(
    rosen_nd, rosenbrock_grad_nd, np.full(10, -1.0)
)
print("10D Conjugate Gradient")
print(f"  Best (first 4): {best_cg10d[:4].round(5)},  f={rosen_nd(best_cg10d):.2e}")
print(f"  Iterations: {iters_cg10d},  Fevals: {feval_cg10d}")
```

## 1.5 — CMA-ES

$$\mathbf{x}_k^{(i)} \sim \mathbf{m} + \sigma \mathcal{N}(\mathbf{0}, \mathbf{C}), \quad i = 1 \ldots \lambda$$
I used the cma package to run this.
Used these tolerances: (`tolfun=1e-11`, `tolx=1e-11`).

```python

def run_cmaes(func, x0, sigma0=0.5):
    opts = cma.CMAOptions()
    opts['tolfun']  = 1e-11
    opts['tolx']    = 1e-11
    opts['verbose'] = -9   # suppress output
    opts['maxfevals'] = 1_000_000

    es = cma.CMAEvolutionStrategy(x0, sigma0, opts)
    path = [np.array(x0)]

    while not es.stop():
        solutions = es.ask()
        es.tell(solutions, [func(s) for s in solutions])
        path.append(es.result.xbest.copy())

    res = es.result
    return res.xbest, res.iterations, res.evaluations, np.array(path)


# 2D
best_cma2d, iters_cma2d, feval_cma2d, path_cma2d = run_cmaes(f2d, [-1.0, -1.0])
print("2D CMA-ES")
print(f"  Best: {best_cma2d.round(6)},  f={f2d(best_cma2d):.2e}")
print(f"  Generations: {iters_cma2d},  Fevals: {feval_cma2d}")

print()

# 10D
best_cma10d, iters_cma10d, feval_cma10d, path_cma10d = run_cmaes(
    rosen_nd, np.full(10, -1.0)
)
print("10D CMA-ES")
print(f"  Best (first 4): {best_cma10d[:4].round(6)},  f={rosen_nd(best_cma10d):.2e}")
print(f"  Generations: {iters_cma10d},  Fevals: {feval_cma10d}")
```

## Testing Jax vs. numpy based differentiation speeds

```python
import numpy as np
import jax
import jax.numpy as jnp
from jax import grad, jit, hessian, vmap
import scipy.optimize as so
import plotly.graph_objects as go
import plotly.io as pio
import time

pio.templates.default = 'plotly_white'

jax.config.update('jax_enable_x64', True)
```

```python
D = 1000
x0_np  = np.full(D, -1.0)
x0_jax = jnp.full(D, -1.0)
def rosen_np(x):
    return np.sum((1 - x[:-1])**2 + 100*(x[1:] - x[:-1]**2)**2)
def rosen_jax(x):
    return jnp.sum((1 - x[:-1])**2 + 100*(x[1:] - x[:-1]**2)**2)
rosen_jit    = jit(rosen_jax)
grad_jit     = jit(grad(rosen_jax))
_ = rosen_jit(x0_jax).block_until_ready()
_ = grad_jit(x0_jax).block_until_ready()
print(f'nupmy = {rosen_np(x0_np):.4f}')
print(f'JAX   = {rosen_jit(x0_jax):.4f}')
```

```python
def finite_diff_grad(f, x, h=1e-5):
    """Forward-difference gradient: D+1 function evaluations."""
    g = np.zeros_like(x)
    f0 = f(x)
    for i in range(len(x)):
        xh = x.copy(); xh[i] += h
        g[i] = (f(xh) - f0) / h
    return g

dims    = [10, 50, 100, 250, 500, 1000, 2000, 5000]
t_fd    = []
t_jax   = []
n_reps  = 5

for d in dims:
    x_np  = np.full(d, -1.0)
    x_jx  = jnp.full(d, -1.0)

    def rosen_d_np(x): return np.sum((1-x[:-1])**2 + 100*(x[1:]-x[:-1]**2)**2)
    def rosen_d_jax(x): return jnp.sum((1-x[:-1])**2 + 100*(x[1:]-x[:-1]**2)**2)

    g_jit_d = jit(grad(rosen_d_jax))
    _ = g_jit_d(x_jx).block_until_ready()  # warm up
    t0 = time.perf_counter()
    for _ in range(n_reps):
        finite_diff_grad(rosen_d_np, x_np)
    t_fd.append((time.perf_counter() - t0) / n_reps * 1000)   # ms

    t0 = time.perf_counter()
    for _ in range(n_reps):
        g_jit_d(x_jx).block_until_ready()
    t_jax.append((time.perf_counter() - t0) / n_reps * 1000)  # ms

    print(f'D={d:5d}  FD: {t_fd[-1]:8.2f} ms   JAX: {t_jax[-1]:6.2f} ms')
```

```python
def gradient_descent_timed(func, grad_fn, x0, tol=1e-5, max_iter=3000):
    x = x0.copy()
    history_f, history_t = [func(x)], [0.0]
    t_start = time.perf_counter()
    f_cur = func(x)

    for k in range(max_iter):
        g = grad_fn(x)
        gn = np.dot(g, g)
        if np.sqrt(gn) < tol:
            break

        alpha = 1e-3
        for _ in range(50):
            x_new = x - alpha * g
            f_new = func(x_new)
            if f_new <= f_cur - 1e-4 * alpha * gn:
                break
            alpha *= 0.5

        x = x_new
        f_cur = f_new
        history_f.append(float(f_cur))
        history_t.append(time.perf_counter() - t_start)

    return x, np.array(history_f), np.array(history_t)


# Wrappers that keep everything in NumPy
def rosen_np_call(x): return float(rosen_np(x))
def grad_jax_np(x):   return np.array(grad_jit(jnp.array(x)))
def grad_fd_np(x):    return finite_diff_grad(rosen_np_call, x, h=1e-5)

print('Running GD with JAX gradient...')
t0 = time.perf_counter()
best_jax, fhist_jax, thist_jax = gradient_descent_timed(
    rosen_np_call, grad_jax_np, x0_np.copy()
)
print(f'  Done in {time.perf_counter()-t0:.2f}s   f={fhist_jax[-1]:.4f}   iters={len(fhist_jax)-1}')

print('Running GD with finite-difference gradient...')
t0 = time.perf_counter()
best_fd, fhist_fd, thist_fd = gradient_descent_timed(
    rosen_np_call, grad_fd_np, x0_np.copy()
)
print(f'  Done in {time.perf_counter()-t0:.2f}s   f={fhist_fd[-1]:.4f}   iters={len(fhist_fd)-1}')
```

```python
hess_jit = jit(hessian(rosen_jax))

def newton_jax(x0, tol=1e-8, max_iter=200):
    x = jnp.array(x0)
    history_f, history_t = [float(rosen_jit(x))], [0.0]
    t_start = time.perf_counter()

    for k in range(max_iter):
        g = grad_jit(x)
        if float(jnp.linalg.norm(g)) < tol:
            break
        H = hess_jit(x)                        
        lam = 1e-4 * float(jnp.trace(H)) / len(x)
        H_reg = H + lam * jnp.eye(len(x))
        d = jnp.linalg.solve(H_reg, g)          
        alpha = 1.0
        f_cur = rosen_jit(x)
        slope = -float(jnp.dot(g, d))
        for _ in range(30):
            x_new = x - alpha * d
            if float(rosen_jit(x_new)) <= float(f_cur) + 1e-4 * alpha * slope:
                break
            alpha *= 0.5

        x = x - alpha * d
        history_f.append(float(rosen_jit(x)))
        history_t.append(time.perf_counter() - t_start)

    return np.array(x), np.array(history_f), np.array(history_t), k+1


t0 = time.perf_counter()
best_newt, fhist_newt, thist_newt, iters_newt = newton_jax(x0_np)
elapsed_newt = time.perf_counter() - t0

print(f'  Converged in {iters_newt} steps')
print(f'  time: {elapsed_newt:.2f}s')
print(f'  f(x*)     : {fhist_newt[-1]:.2e}')
print(f'  x* (first 4): {best_newt[:4].round(5)}')
```
